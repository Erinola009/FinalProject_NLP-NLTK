# Step 1: Understanding the Purpose of Each NLP Technique
### let's briefly understand each NLP technique:

* Tokenization – Splitting text into words or sentences.

* Stopword Removal – Removing common words like "the," "is," "in" that do not add much meaning.

* POS (Part-of-Speech) Tagging – Identifying nouns, verbs, adjectives, etc.

* Lemmatization – Reducing words to their base form (e.g., "running" → "run").

* Stemming – Cutting words down to their root (e.g., "studies" → "studi").

* Named Entity Recognition (NER) – Identifying names of people, places, dates, etc.

* TF-IDF (Term Frequency - Inverse Document Frequency) – Measuring the importance of words in a document.

* Sentiment Analysis – Detecting emotions (positive, negative, neutral).

* Text Classification – Assigning text to categories (e.g., "spam" or "not spam").

```python
#Text Classification (Spam Detection Example)
#Import your libraries
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# MySample Dataset
emails = ["Win a Sporty now!", "Your invoice is due", "Play Now Get rich fast!", "Meeting scheduled tomorrow"]
labels = [1, 0, 1, 0]  # 1 = spam, 0 = not spam

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)
classifier = MultinomialNB()
classifier.fit(X, labels)

# Test new email
new_email = ["Exclusive deal, claim now!"]
X_test = vectorizer.transform(new_email)
prediction = classifier.predict(X_test)

print("Spam or Not:", "Spam" if prediction[0] == 1 else "Not Spam")
```

# Step 3: Evaluating NLP Techniques on a Dataset
##### Now that we have implemented NLP techniques, let's evaluate their performance using a dataset. We'll analyze:
* Checking accuracy using test data.
* Comparing lemmatization vs. stemming accuracy.
* Measuring TF-IDF scores for word importance.
* Testing sentiment analysis results on different text samples

```python
#1. Checking Accuracy Using Test Data
#We'll use a Naive Bayes classifier to classify text into two categories (e.g., spam vs. not spam). Then, we'll evaluate its accuracy using test data.

#Dataset (Example: Spam Detection)
#We'll create a small dataset where spam messages contain promotional content.

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Sample dataset
emails = ["Win free 8 months course ML in GoMYCode Academy before that you need to click the link above", 
    "Win a lottery now!", "Your invoice is due", "Get rich fast!", "Meeting scheduled tomorrow", "Congratulations, you've won!", "Earn money quickly"]
labels = [1, 0, 1, 0, 1, 0, 1]  # 1 = spam, 0 = not spam

# Splitting into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(emails, labels, test_size=0.3, random_state=42)

# Vectorizing the text
vectorizer = CountVectorizer()
X_train_vect = vectorizer.fit_transform(X_train)
X_test_vect = vectorizer.transform(X_test)

# Training the Naive Bayes classifier
classifier = MultinomialNB()
classifier.fit(X_train_vect, y_train)

# Predictions and accuracy check
y_pred = classifier.predict(X_test_vect)
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
#2. Comparing Lemmatization vs. Stemming Accuracy
#To see how lemmatization and stemming affect classification accuracy, we'll process the dataset differently and compare results.

from nltk.stem import WordNetLemmatizer, PorterStemmer
from nltk.tokenize import word_tokenize
import nltk

lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()

# MyExample words for comparison
words = ["running", "flies", "better", "driving", "easily"]

# Applying Lemmatization
lemmatized_words = [lemmatizer.lemmatize(word, pos='v') for word in words]

# Applying Stemming
stemmed_words = [stemmer.stem(word) for word in words]

print("Original Words:", words)
print("Lemmatized Words:", lemmatized_words)
print("Stemmed Words:", stemmed_words)
```

```python
#Analysis:
#Lemmatization produces actual words that maintain meaning.
#Stemming sometimes cuts words incorrectly ("flies" → "fli", "easily" → "easili").
#Lemmatization is usually better for accuracy in classification.
```

```python
#3. Measuring TF-IDF Scores for Word Importance
#We'll apply TF-IDF (Term Frequency-Inverse Document Frequency) to measure word importance.

from sklearn.feature_extraction.text import TfidfVectorizer

# MySample text corpus
Text_documents = ["NLP is amazing and useful", "Machine learning is the future", 
             "Deep learning and NLP work together", "NLP applications are growing"]

# Applying TF-IDF
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(Text_documents)

# Display TF-IDF scores
feature_names = vectorizer.get_feature_names_out()
for doc_idx, doc in enumerate(X.toarray()):
    print(f"Document {doc_idx + 1}:")
    for word_idx, score in enumerate(doc):
        if score > 0:  # Display only important words
            print(f"  {feature_names[word_idx]}: {score:.3f}")
```

```python
#Analysis:
#Words that appear frequently but in fewer documents (eample,  "amazing", "future") have higher scores.
#Common words like "is" have lower scores because they appear in multiple documents.
#TF-IDF helps determine which words are most important for text classification.
```

```python
#4. Testing Sentiment Analysis on Different Text Samples
#We'll check how sentiment analysis classifies positive, negative, and neutral sentences.

from textblob import TextBlob

texts = [
    "I love this DS Course! It's fantastic.",
    "Studying at GoMyCode is Best Choice I've ever Made.",
    "The movie was okay, not the best but not the worst.",
    "I absolutely hate how this turned out.",
    "The service was good, but the waiting time was too long."
]

# Analyzing sentiment
for text in texts:
    sentiment_score = TextBlob(text).sentiment.polarity
    sentiment = "Positive" if sentiment_score > 0 else "Negative" if sentiment_score < 0 else "Neutral"
    print(f"Text: {text}\nSentiment Score: {sentiment_score:.2f} → {sentiment}\n")
```

```python
#Analysis:
#Sentiment analysis correctly identifies positive, negative, and neutral tones.
#It can be useful for product reviews, social media monitoring, and customer feedback analysis.
```
